```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as blt
```

```python
a=pd.read_csv(r"Downloads/insurance.csv")
a
```

```python
## 1.create a line plot showing age versus insurance charges:
age_charges=a.groupby('age')['charges'].mean()
plt.figure(figsize=(10,5))
plt.plot(age_charges.index,age_charges.values,marker='o',linestyle='-')
plt.title("Average Insurance Charges by Age")
plt.xlabel("Age")
plt.ylabel("Average Insurance Charges ($)")
plt.grid(True)
plt.show()
```

```python
## 2.create a bar chart showing customer count by region:
region_count = a['region'].value_counts()
plt.figure(figsize=(8, 5))
plt.bar(region_count.index, region_count.values)
plt.title("Customer Count by Region")
plt.xlabel("Region")
plt.ylabel("Customer Count")
plt.grid(axis='y')
plt.show()
     
```

```python
## 3.create a pie charts showing smoker versus non-smoker distribution:
smoker_count = a['smoker'].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(smoker_count.values,labels=smoker_count.index,autopct='%1.1f%%',startangle=90)

plt.title("Smoker Distribution")
plt.show()
     
```

```python
## 3. creat a histogram of customer ages:
plt.figure(figsize=(8, 5))
plt.hist(a['age'], bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.grid(axis='y')
plt.show()
     
```

```python
## 4.create a histogram of insurance ages:
plt.figure(figsize=(8, 5))
plt.hist(a['charges'], bins=20)
plt.title("Insurance Charges Distribution")
plt.xlabel("Insurance Charges")
plt.ylabel("Frequency")
plt.grid(axis='y')
plt.show()
     
```

```python
##5.create a scotter plot of bmi charges:
plt.figure(figsize=(8, 5))
plt.scatter(a['bmi'], a['charges'])
plt.title("BMI vs Insurance Charges")
plt.xlabel("BMI")
plt.ylabel("Insurance Charges")
plt.grid(True)
plt.show()
     
```

```python
## 6. create a horizontal bar chart  showing average charges by gender:
gender_charges = a.groupby('sex')['charges'].mean()
plt.figure(figsize=(8, 5))
plt.barh(gender_charges.index, gender_charges.values)
plt.title("Average Charges by Gender")
plt.xlabel("Average Charges")
plt.ylabel("Gender")

plt.grid(axis='x')
plt.show()
```

```python
## 7.Create a box plot showing charges distribution.
plt.figure(figsize=(8, 5))
plt.boxplot(a['charges'])
plt.title("Insurance Charges Distribution")
plt.ylabel("Insurance Charges")
plt.grid(axis='y')
plt.show()
     
```

```python
##8.Compare smoker and non-smoker charges using a bar chart.
smoker_charges = a.groupby('smoker')['charges'].mean()
plt.figure(figsize=(8, 5))
plt.bar(smoker_charges.index, smoker_charges.values)
plt.title("Average Charges: Smokers vs Non-Smokers")
plt.xlabel("Smoker Status")
plt.ylabel("Average Charges")
plt.grid(axis='y')
plt.show()
```

```python
## 9.Create separate histograms for male and female customers.
male = a[a['sex'] == 'male']
female = a[a['sex'] == 'female']
plt.figure(figsize=(10, 5))
plt.hist(male['age'], bins=10,alpha=0.5, label='Male')
plt.hist(female['age'], bins=10,alpha=0.5, label='Female')
plt.title("Age Distribution by Gender")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.legend()
plt.grid(axis='y')
plt.show()
     
```

```python
## 10.Visualize average charges by region.
region_charges = a.groupby('region')['charges'].mean()
plt.figure(figsize=(9, 5))
plt.bar(region_charges.index, region_charges.values)
plt.title("Average Insurance Charges by Region")
plt.xlabel("Region")
plt.ylabel("Average Charges ($)")
plt.grid(axis='y')
plt.show()
```
