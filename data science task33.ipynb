```python
i['order_purchase_timestamp'] = pd.to_datetime(i['order_purchase_timestamp'])
print(i['order_purchase_timestamp'].dtype)
```

```python
i['year'] = i['order_purchase_timestamp'].dt.year
i['month'] = i['order_purchase_timestamp'].dt.month
i['day'] = i['order_purchase_timestamp'].dt.day
print(i.head())
```

```python
i['order_delivered_customer_date'] = pd.to_datetime(i['order_delivered_customer_date'])
i['delivery_duration'] = (i['order_delivered_customer_date'] - i['order_purchase_timestamp']).dt.days
print(i[['order_id', 'delivery_duration']])
```

```python
i['order_estimated_delivery_date'] = pd.to_datetime(i['order_estimated_delivery_date'])
i['delivery_difference'] = (i['order_delivered_customer_date'] - i['order_estimated_delivery_date']).dt.days
print(i[['order_id', 'delivery_difference']])
```

```python
def check_status(days):
    if pd.isna(days):
        return 'Not Delivered'
    elif days < 0:
        return 'Early'
    elif days == 0:
        return 'On Time'
    else:
        return 'Delayed'

i['delivery_status'] = i['delivery_difference'].apply(check_status)

print(i[['order_id', 'delivery_status']])
```
