```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv(r"Downloads/insurance.csv")
df
```

```python
# 1.Convert the age column into a NumPy array.
age = df['age'].to_numpy()
print(age)
```

```python
#2.Convert the BMI column into a NumPy array.
bmi = df['bmi'].to_numpy()
print(bmi)
```

```python
#3.Extract the insurance charges into an array.
charges = df['charges'].to_numpy()
print(charges)
```

```python
#4.Find the total number of customers.
print(len(age))
     
```

```python
#5.Find the average age.
avg_age = np.mean(age)
print("Average age: ",avg_age)
```

```python
#6.Find the minimum and maximum BMI.
print("Minimum BMI: ", np.min(bmi))
print("Maximum BMI: ", np.max(bmi))
```

```python
#7.Calculate the average insurance charge.
avg_charge = np.mean(charges)
print("Average Insurance Charge: ",avg_charge)
     
```

```python
#8.Find the median insurance charge.
median_charge = np.median(charges)
print("Median Insurance Charge: ",median_charge)
     
```

```python
#9.Calculate the standard deviation of charges.
std_charge = np.std(charges)
print("Standard Deviation of charges: ", std_charge)
```

```python
#10.Count how many customers are above 40 years old.
cust = np.sum(age>40)
print("Customers above age 40: ",cust)
```

```python
#11.Find customers whose insurance charges are greater than ₹20,000 (dataset charges are in USD, so use $20,000).
high_charges= charges[charges > 20000]
print(high_charges)
print(len(high_charges))
     
```

```python
#12.Calculate the 25th, 50th and 75th percentiles of insurance charges.
percentile_25 = np.percentile(charges, 25)
percentile_50 = np.percentile(charges, 50)
percentile_75 = np.percentile(charges, 75)
print("Q1: ",percentile_25)
print("Q2: ",percentile_50)
print("Q3: ",percentile_75)
```

```python
#13.Calculate the variance of insurance charges.
variance_charges = np.var(charges)
print("Variance of charges: ",variance_charges)
```

```python
#14.Normalize the age array using min-max normalization.
normalized_age = (age - np.min(age)) / (np.max(age) - np.min(age))
print("Normalized age: ",normalized_age)
```

```python
#15.Calculate the difference between maximum and minimum charges.
charge_diff = np.max(charges) - np.min(charges)
print("Charge Difference: ", charge_diff)
```

```python
#16.Find the indices of the five customers with the highest insurance charges.
top_indices = np.argsort(charges)[-5:][::-1]
print("High Charge customers: ",top_indices)
```

```python
#17.Replace all BMI values above 40 with the mean BMI for a separate analytical array.
mean_bmi = np.mean(bmi)
mod_bmi = np.where(bmi > 40, mean_bmi, bmi)
print("Modified BMI: ", mod_bmi)
```

```python
#18.Calculate the correlation between age and charges using NumPy.
correlation = np.corrcoef(age, charges)
print(correlation)
print("Correlation coefficient: ",correlation[0,1])
```

```python
#19.Calculate average charges by gender and smoker status.
gender_charges = df.groupby('sex')['charges'].mean()
print("Average charge by gender: ")
print(gender_charges)
result = df.groupby(['sex', 'smoker'])['charges'].mean()
print("Average charge by gender and cmoker status: ")
print(result)
     
```

```python
#20.Find the top five customers with the highest charges.
top_five = df.nlargest(5, 'charges')
print("Top 5 customers by charges")
print(top_five)
     
```

```python

#21.Find the region with the highest average insurance charges.
regions = df.groupby('region')['charges'].mean()
print("Region with the highest average charge")
print(regions.sort_values(ascending = False))
```

```python
#22.Compare the average charges of smokers and non-smokers.
smoker_charges = df.groupby('smoker')['charges'].mean()
print(smoker_charges)
     
```

```python
#Find the average charges for each combination of region and gender.
region_gender = df.groupby(['region', 'sex'])['charges'].mean()
print(region_gender)
     
```
